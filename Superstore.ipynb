Mərhələ 4. SQL ilə Analiz və Sorğular

```python
import sqlite3

# 1. Yaddaşda (in-memory) müvəqqəti SQL bazası yaradırıq
conn = sqlite3.connect(':memory:')

# 2. Pandas dataframe-imizi 'superstore' adlı SQL cədvəlinə çeviririk
df.to_sql('superstore', conn, index=False, if_exists='replace')

print("Data uğurla SQL bazasına yükləndi!")
```

```python
# 1. SQL sorğusunu yazırıq
query = """
SELECT Category, [Sub-Category], Profit
FROM superstore
LIMIT 5;
"""

# 2. Nəticəni oxuyuruq
result = pd.read_sql_query(query, conn)
print(result)
```

```python
# 1. SQL sorğusu: Kateqoriyalar üzrə mənfəəti toplayırıq
query = """
SELECT Category, SUM(Profit) AS Total_Profit
FROM superstore
GROUP BY Category
ORDER BY Total_Profit ASC;
"""

# 2. Sorğunu icra edirik
result = pd.read_sql_query(query, conn)
print(result)
```

```python
# 1. SQL sorğusunu yazırıq
query = """
SELECT [Sub-Category], SUM(Profit) AS Total_Profit
FROM superstore
GROUP BY [Sub-Category]
ORDER BY Total_Profit ASC;
"""

# 2. Sorğunu icra edib nəticəni pandas dataframe kimi oxuyuruq
result = pd.read_sql_query(query, conn)
print(result)
```

```python
# 1. SQL sorğusunu yazırıq (WHERE ilə sətirləri filtrləyirik)
query = """
SELECT [Order ID], Sales, Discount, Profit
FROM superstore
WHERE Discount > 0.2
LIMIT 5;
"""

# 2. Sorğunu icra edib nəticəni oxuyuruq
result = pd.read_sql_query(query, conn)
print(result)
```

```python
# 1. SQL sorğusunu yazırıq (WHERE ilə sətirləri filtrləyirik)
query = """
SELECT [Order ID], [Sub-Category], Sales, Profit
FROM superstore
WHERE Profit < 0
LIMIT 10;
"""

# 2. Sorğunu icra edib nəticəni oxuyuruq
result = pd.read_sql_query(query, conn)
print(result)
```

```python
# 1. SQL sorğusunu yazırıq (HAVING ilə cəmlənmiş nəticəni filtrləyirik)
query = """
SELECT [Sub-Category], SUM(Profit) AS Total_Profit
FROM superstore
GROUP BY [Sub-Category]
HAVING SUM(Profit) > 30000
ORDER BY Total_Profit DESC;
"""

# 2. Sorğunu icra edib nəticəni oxuyuruq
result = pd.read_sql_query(query, conn)
print(result)
```

```python
# 1. SQL sorğusunu yazırıq (HAVING ilə cəmlənmiş nəticəni filtrləyirik)
query = """
SELECT [Sub-Category], SUM(Profit) AS Total_Profit
FROM superstore
GROUP BY [Sub-Category]
HAVING SUM(Profit) < 0
ORDER BY Total_Profit ASC;
"""

# 2. Sorğunu icra edib nəticəni oxuyuruq
result = pd.read_sql_query(query, conn)
print(result)
```

```python
# 1. Sub-Category üzrə Sales və Profit cəmini tapırıq
subcat_summary = df.groupby('Sub-Category')[['Sales', 'Profit']].sum().reset_index()

# 2. Mənfəətə görə en azdan en çox doğru sıralayırıq
subcat_summary = subcat_summary.sort_values(by='Profit', ascending=True)

# 3. Nəticəni ekrana çıxarırıq
print(subcat_summary)
```

```python
# Region üzrə Sales və Profit cəmini tapıb mənfəətə görə sıralayırıq
region_summary = df.groupby('Region')[['Sales', 'Profit']].sum().reset_index()
region_summary = region_summary.sort_values(by='Profit', ascending=True)

print(region_summary)
```
